### Обучение модели

```python
import os
import glob
import random
import numpy as np
import pandas as pd
import cv2
from PIL import Image, ImageOps, ImageFile
from tqdm import tqdm
from sklearn.model_selection import train_test_split

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as T
import timm

ImageFile.LOAD_TRUNCATED_IMAGES = True
```

```python
SEED = 42
BATCH_SIZE = 64
NUM_EPOCHS = 15
LR = 3e-4
TARGET_SIZE = (256, 64)  # (Ширина, Высота)
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

seed_everything(SEED)

VALID_EXTENSIONS = {'.jpg', '.jpeg', '.png', '.bmp', '.webp', '.tiff'}

def get_image_paths(directory):
    """Рекурсивно собирает валидные пути к изображениям"""
    image_paths = []
    if not os.path.exists(directory):
        print(f"Предупреждение: Путь {directory} не найден!")
        return image_paths

    for root, _, files in os.walk(directory):
        for file in files:
            ext = os.path.splitext(file)[1].lower()
            if ext in VALID_EXTENSIONS:
                image_paths.append(os.path.join(root, file))
    return image_paths
```

```python
def resize_and_pad_smart(img: Image.Image, target_size=(256, 64)):
    """
    Масштабирует изображение с сохранением пропорций.
    Паддинг заполняется дублированием краевых пикселей (BORDER_REPLICATE)
    вместо искусственного серого фона, устраняя искусственные границы.
    """
    target_w, target_h = target_size
    w, h = img.size
    if w == 0 or h == 0:
        return Image.new('RGB', target_size, (127, 127, 127))

    scale = min(target_w / w, target_h / h)
    new_w, new_h = max(1, int(round(w * scale))), max(1, int(round(h * scale)))

    img_np = np.array(img)
    img_resized = cv2.resize(img_np, (new_w, new_h), interpolation=cv2.INTER_LINEAR)

    pad_w = target_w - new_w
    pad_h = target_h - new_h

    top = pad_h // 2
    bottom = pad_h - top
    left = pad_w // 2
    right = pad_w - left

    img_padded = cv2.copyMakeBorder(
        img_resized, top, bottom, left, right,
        borderType=cv2.BORDER_REPLICATE
    )
    return Image.fromarray(img_padded)

# Нормализация ImageNet
normalize_transform = T.Compose([
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
```

```python
class OrientationDataset(Dataset):
    def __init__(self, data_list, target_size=(256, 64), is_train=True):
        self.data_list = data_list
        self.target_size = target_size
        self.is_train = is_train

    def __len__(self):
        return len(self.data_list)

    def __getitem__(self, idx):
        img_path, _ = self.data_list[idx]

        try:
            img = Image.open(img_path).convert('RGB')
        except Exception:
            img = Image.new('RGB', self.target_size, (127, 127, 127))

        label = 0.0
        if self.is_train and random.random() < 0.5:
            img = img.rotate(180)
            label = 1.0

        img = resize_and_pad_smart(img, target_size=self.target_size)
        tensor_img = normalize_transform(img)
        return tensor_img, torch.tensor(label, dtype=torch.float32)


class TestDatasetTTA(Dataset):
    def __init__(self, image_paths, target_size=(256, 64)):
        self.image_paths = image_paths
        self.target_size = target_size

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        path = self.image_paths[idx]
        try:
            img = Image.open(path).convert('RGB')
        except Exception:
            img = Image.new('RGB', self.target_size, (127, 127, 127))

        # 1. Исходный кроп
        img_0 = resize_and_pad_smart(img, target_size=self.target_size)

        # 2. Кроп, повернутый на 180°
        img_180 = resize_and_pad_smart(img.rotate(180), target_size=self.target_size)

        filename = os.path.basename(path)
        image_id = os.path.splitext(filename)[0]

        return normalize_transform(img_0), normalize_transform(img_180), image_id
```

```python
class BCEMSELoss(nn.Module):
    """
    Комбинация BCELoss и MSELoss. Оптимизирует калибровку вероятностей
    напрямую под метрику Brier Score (MSE между вероятностью и классом).
    """
    def __init__(self, alpha=0.6):
        super().__init__()
        self.bce = nn.BCEWithLogitsLoss()
        self.mse = nn.MSELoss()
        self.alpha = alpha

    def forward(self, logits, targets):
        probs = torch.sigmoid(logits)
        loss_bce = self.bce(logits, targets)
        loss_mse = self.mse(probs, targets)
        return self.alpha * loss_bce + (1.0 - self.alpha) * loss_mse
```

```python
real_paths = (
    get_image_paths('/content/english_high_quality/IIIT5K-Word_V3.0/IIIT5K/train') +
    get_image_paths('/content/english_high_quality/IIIT5K-Word_V3.0/IIIT5K/test') +
    get_image_paths('/content/russian_cursive/train') +
    get_image_paths('/content/russian_cursive/test')
)

synthetic_paths = get_image_paths('/content/synthetic_dataset/images')

print(f"Найдено реальных картинок: {len(real_paths)}")
print(f"Найдено синтетических картинок: {len(synthetic_paths)}")

data = [(p, False) for p in real_paths] + [(p, True) for p in synthetic_paths]

train_data, val_data = train_test_split(data, test_size=0.1, random_state=SEED, shuffle=True)

test_paths = get_image_paths('/content/test_images/images')
if not test_paths:
    test_paths = get_image_paths('/content/test_images')

print(f"Размер Train: {len(train_data)}, Val: {len(val_data)}, Test: {len(test_paths)}")

train_dataset = OrientationDataset(train_data, target_size=TARGET_SIZE, is_train=True)
val_dataset = OrientationDataset(val_data, target_size=TARGET_SIZE, is_train=False)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=4, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=4, pin_memory=True)

model = timm.create_model('convnext_tiny.fb_in22k_ft_in1k', pretrained=True, num_classes=1)
model = model.to(DEVICE)

criterion = BCEMSELoss(alpha=0.6)
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-2)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS, eta_min=1e-6)
```

```python
print(f"\n--- Старт обучения ConvNeXt на {DEVICE} ---")

best_brier = 0.0

for epoch in range(1, NUM_EPOCHS + 1):
    # Обучение
    model.train()
    running_loss = 0.0
    for images, labels in tqdm(train_loader, desc=f"Epoch {epoch}/{NUM_EPOCHS} [Train]"):
        images = images.to(DEVICE)
        labels = labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(images).squeeze(-1)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)

    scheduler.step()
    train_loss = running_loss / len(train_dataset)

    # Валидация
    model.eval()
    val_loss = 0.0
    all_preds = []
    all_targets = []

    with torch.no_grad():
        for images, labels in tqdm(val_loader, desc=f"Epoch {epoch}/{NUM_EPOCHS} [Val]"):
            images = images.to(DEVICE)
            labels = labels.to(DEVICE)

            outputs = model(images).squeeze(-1)
            loss = criterion(outputs, labels)
            val_loss += loss.item() * images.size(0)

            preds = torch.sigmoid(outputs)
            all_preds.extend(preds.cpu().numpy())
            all_targets.extend(labels.cpu().numpy())

    val_loss = val_loss / len(val_dataset)
    brier_score = np.mean((np.array(all_preds) - np.array(all_targets)) ** 2)
    one_minus_brier = 1.0 - brier_score

    print(f"Эпоха {epoch:02d} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | Val (1 - Brier Score): {one_minus_brier:.5f}")

    if one_minus_brier > best_brier:
        best_brier = one_minus_brier
        torch.save(model.state_dict(), "best_model.pth")

print(f"\nЛучший результат на валидации (1 - Brier Score): {best_brier:.5f}")

```

```python
print("\n--- Загрузка лучшей модели и генерация предсказаний с TTA ---")
model.load_state_dict(torch.load("best_model.pth"))
model.eval()

test_dataset = TestDatasetTTA(test_paths, target_size=TARGET_SIZE)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=4)

results = []

with torch.no_grad():
    for img_0, img_180, image_ids in tqdm(test_loader, desc="Predicting Test (TTA)"):
        img_0 = img_0.to(DEVICE)
        img_180 = img_180.to(DEVICE)

        logits_0 = model(img_0).squeeze(-1)
        logits_180 = model(img_180).squeeze(-1)

        p0 = torch.sigmoid(logits_0).cpu().numpy()
        p180 = torch.sigmoid(logits_180).cpu().numpy()

        p_final = (p0 + (1.0 - p180)) / 2.0

        for img_id, p in zip(image_ids, p_final):
            p_clipped = float(np.clip(p, 0.0001, 0.9999))
            results.append({"image_id": img_id, "p_180": p_clipped})

df_sub = pd.DataFrame(results)
df_sub = df_sub.sort_values(by="image_id", ascending=True)

output_path = "submission.csv"
df_sub.to_csv(output_path, index=False)
print(f"\nФайл предсказаний успешно сохранен в '{output_path}'. Строк: {len(df_sub)}")
```

```python
'''
import os
import random
import string
import glob
import numpy as np
import pandas as pd
import cv2
from PIL import Image, ImageDraw, ImageFont
import albumentations as A

random.seed(42)
np.random.seed(42)

RU_WORDS = [
    # Категории и продажи
    "Продам", "Состояние", "Отличное", "Новый", "Бесплатно", "Доставка", "Авито",
    "Гарантия", "В наличии", "Оригинал", "Скидка", "Москва", "Ремонт", "Купить",
    "Срочно", "Цена", "Рублей", "Услуги", "Телефон", "Ноутбук", "Запчасти", "Аренда",
    "Паспорт", "Договор", "Автомобиль", "Квартира", "Комплект", "Размер", "Производство",
    "Магазин", "Оптом", "Розница", "Штука", "Сервис", "Обмен", "Торг", "Самовывоз",
    "Коляска", "Шины", "Диски", "Одежда", "Обувь", "Мебель", "Спецтехника", "Работа",
    "Вакансия", "Оплата", "Карта", "Проверка", "Чек", "Гарантийный", "Талон", "Сборка",
    "Установка", "Официальный", "Дилер", "Без пробега", "Б/У", "Отличный", "Идеальный",
    "Качество", "Фирменный", "Сертифицирован", "Акция", "Подарок", "Рассрочка", "Кредит",
    "Выгода", "Спешите", "Количество", "Ограничено", "Комплектация", "Модель", "Серия",
    "Артикул", "Штрихкод", "Адрес", "Контакт", "Номер", "Внимание", "Распродажа",
    # Описание состояния и параметров
    "Полный комплект", "Не бит", "Не крашен", "Рабочий", "Исправен", "Требуется ремонт",
    "На запчасти", "Сборка Россия", "Импортный", "Кожаный", "Деревянный", "Металлический",
    "Пластиковый", "Белый", "Черный", "Серый", "Красный", "Синий", "Зеленый", "Бежевый"
]

EN_WORDS = [
    # Бренды и коммерция
    "Sale", "iPhone", "Samsung", "Nike", "Adidas", "Pro", "Max", "Original",
    "Free", "Delivery", "New", "Used", "Brand", "Size", "Quality", "Price",
    "Discount", "Available", "Auto", "Ultra", "Global", "Version", "Edition",
    "Model", "Serial", "Number", "Made in", "Authentic", "Limited", "Premium",
    "Official", "Store", "Shop", "Special", "Offer", "Order", "Express", "Fast",
    "Shipping", "Guarantee", "Warranty", "Pcs", "Set", "Pack", "Type", "Style",
    "Color", "Black", "White", "Red", "Blue", "Gold", "Silver", "Leather", "Cotton",
    "Metal", "Plastic", "Smart", "Watch", "Wireless", "Bluetooth", "Battery",
    "Charger", "Cable", "Case", "Cover", "Glass", "Plus", "Lite", "Mini", "Air",
    "Master", "Super", "Gaming", "Pro Max", "Standard", "Custom", "Heavy Duty",
    "High Quality", "Top Rated", "Best Seller", "In Stock", "Fast Delivery"
]

PUNCTUATION_SUFFIXES = ["", "", "", ",", ".", "!", "?", "%", " руб.", " шт.", " кг", " USD", " EUR", " грн."]
PUNCTUATION_PREFIXES = ["", "", "", "№", "«", "\"", "(", "#"]


MU_W, SIGMA_W = 5.472, 0.902
MU_H, SIGMA_H = 3.738, 0.924

def sample_crop_size():
    w = int(np.clip(np.random.lognormal(MU_W, SIGMA_W), 28, 1200))
    h = int(np.clip(np.random.lognormal(MU_H, SIGMA_H), 12, 300))
    return w, h


def get_word():
    """Возвращает случайное слово, число или цену с возможными знаками препинания."""
    choice = random.choice(["ru", "en", "num", "price", "code"])
    if choice == "ru":
        word = random.choice(RU_WORDS)
    elif choice == "en":
        word = random.choice(EN_WORDS)
    elif choice == "num":
        word = str(random.randint(1, 99999))
    elif choice == "price":
        word = f"{random.randint(10, 50000)}{random.choice(['₽', ' руб.', ' $', ' €', ''])}"
    else:
        word = f"{random.choice(string.ascii_uppercase)}{random.randint(100, 999)}"

    # Добавляем знаки препинания с вероятностью ~35%
    if random.random() < 0.35:
        prefix = random.choice(PUNCTUATION_PREFIXES)
        suffix = random.choice(PUNCTUATION_SUFFIXES)

        # Парные кавычки/скобки
        if prefix == "«":
            suffix = "»"
        elif prefix == "(":
            suffix = ")"
        elif prefix == "\"":
            suffix = "\""

        word = f"{prefix}{word}{suffix}"

    return word

def apply_arc_distortion(pil_rgba_img, curvature_limit=0.25):
    """
    Применяет изгиб текста по дуге с помощью трансформирующей сетки (cv2.remap).
    """
    img_np = np.array(pil_rgba_img)
    h, w = img_np.shape[:2]
    if h < 10 or w < 20:
        return pil_rgba_img

    # Создаем сетку
    x = np.arange(w, dtype=np.float32)
    y = np.arange(h, dtype=np.float32)
    grid_x, grid_y = np.meshgrid(x, y)

    # Нормализованные X от -1 до 1
    norm_x = (2.0 * grid_x / w) - 1.0

    # Вычисляем деформацию по дуге (параболическую)
    direction = random.choice([-1, 1])
    amplitude = h * random.uniform(0.1, curvature_limit) * direction
    shift_y = amplitude * (1.0 - norm_x**2)

    src_x = grid_x
    src_y = grid_y - shift_y

    # Применяем деформацию
    warped = cv2.remap(
        img_np,
        src_x.astype(np.float32),
        src_y.astype(np.float32),
        interpolation=cv2.INTER_LINEAR,
        borderMode=cv2.BORDER_CONSTANT,
        borderValue=(0, 0, 0, 0)
    )
    return Image.fromarray(warped)

def find_max_font_size(text, font_path, max_w, max_h):
    """Бинарный поиск максимального шрифта, вписывающегося в max_w и max_h."""
    low, high = 6, max(12, max_h * 3)
    best_font = ImageFont.load_default()

    while low <= high:
        mid = (low + high) // 2
        try:
            font = ImageFont.truetype(font_path, size=mid) if font_path else ImageFont.load_default()
        except Exception:
            font = ImageFont.load_default()

        bbox = font.getbbox(text)
        tw, th = bbox[2] - bbox[0], bbox[3] - bbox[1]

        if tw <= max_w and th <= max_h:
            best_font = font
            low = mid + 1
        else:
            high = mid - 1

    return best_font


def render_text_overlay(font_path, target_w, target_h):
    """
    Генерирует текст, подбирает размер шрифта и (опционально) искривляет по дуге,
    гарантируя, что итоговый оверлей поместится вtarget_w х target_h.
    """
    margin_x = max(2, int(target_w * random.uniform(0.02, 0.05)))
    margin_y = max(2, int(target_h * random.uniform(0.02, 0.05)))

    max_w = max(10, target_w - 2 * margin_x)
    max_h = max(8, target_h - 2 * margin_y)

    text_words = [get_word()]

    # Собираем строку из 1-6 слов
    while True:
        text = " ".join(text_words)
        font = find_max_font_size(text, font_path, max_w, max_h)
        bbox = font.getbbox(text)
        tw = bbox[2] - bbox[0]

        if tw >= max_w * 0.75 or len(text_words) >= 6:
            break
        text_words.append(get_word())

    text = " ".join(text_words)
    font = find_max_font_size(text, font_path, max_w, max_h)

    # Отрисовываем текст на временном холсте RGBA
    bbox = font.getbbox(text)
    tw, th = max(1, bbox[2] - bbox[0]), max(1, bbox[3] - bbox[1])

    # Холст с запасом по высоте под возможную дугу
    canvas_h = int(th * 2.5) + 10
    canvas_w = tw + 10
    text_canvas = Image.new("RGBA", (canvas_w, canvas_h), (0, 0, 0, 0))
    draw = ImageDraw.Draw(text_canvas)

    # Рисуем текст в центре холста
    tx = (canvas_w - tw) // 2 - bbox[0]
    ty = (canvas_h - th) // 2 - bbox[1]
    draw.text((tx, ty), text, font=font, fill=(255, 255, 255, 255))

    # С вероятностью 25% делаем изгиб по дуге
    is_curved = random.random() < 0.25
    if is_curved:
        text_canvas = apply_arc_distortion(text_canvas)

    # Кадрируем по видимым пикселям
    bbox_crop = text_canvas.getbbox()
    if bbox_crop:
        text_canvas = text_canvas.crop(bbox_crop)

    # Масштабируем, если после дуги текст вышел за пределы max_w/max_h
    cur_w, cur_h = text_canvas.size
    scale = min(max_w / max(cur_w, 1), max_h / max(cur_h, 1), 1.0)
    if scale < 1.0:
        new_w = max(1, int(cur_w * scale))
        new_h = max(1, int(cur_h * scale))
        text_canvas = text_canvas.resize((new_w, new_h), Image.BICUBIC)

    return text_canvas


def create_background(w, h):
    bg_type = random.choice(["solid", "gradient", "noise"])
    if bg_type == "solid":
        color = random.randint(180, 255) if random.random() < 0.7 else random.randint(0, 80)
        return Image.fromarray(np.full((h, w, 3), color, dtype=np.uint8))
    elif bg_type == "gradient":
        c1 = np.array([random.randint(150, 255) for _ in range(3)])
        c2 = np.array([random.randint(150, 255) for _ in range(3)])
        grad = np.linspace(0, 1, h)[:, None, None]
        bg = (c1 * (1 - grad) + c2 * grad).astype(np.uint8)
        return Image.fromarray(np.tile(bg, (1, w, 1)))
    else:
        noise = np.random.normal(200, 20, (h, w, 3))
        return Image.fromarray(np.clip(noise, 0, 255).astype(np.uint8))

def get_contrasting_color(bg_img):
    mean_val = np.mean(np.array(bg_img))
    if mean_val > 127:
        return (random.randint(0, 50), random.randint(0, 50), random.randint(0, 50))
    return (random.randint(200, 255), random.randint(200, 255), random.randint(200, 255))


def get_augmentation_pipeline():
    return A.Compose([
        A.OneOf([
            A.GaussianBlur(blur_limit=(3, 5), p=0.4),
            A.MotionBlur(blur_limit=(3, 5), p=0.4),
        ], p=0.4),
        A.GaussNoise(var_limit=(10.0, 40.0), p=0.3),
        A.RandomBrightnessContrast(brightness_limit=0.15, contrast_limit=0.15, p=0.4),
        A.ImageCompression(quality_lower=40, quality_upper=90, p=0.5),
        A.Perspective(scale=(0.01, 0.03), p=0.2),
    ])


class SyntheticDatasetGenerator:
    def __init__(self, fonts_dir=None):
        self.fonts = []
        if fonts_dir and os.path.exists(fonts_dir):
            self.fonts = glob.glob(os.path.join(fonts_dir, "*.ttf")) + glob.glob(os.path.join(fonts_dir, "*.otf"))

        if not self.fonts:
            system_fonts = [
                "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf",
                "/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf",
                "C:\\Windows\\Fonts\\arial.ttf",
                "C:\\Windows\\Fonts\\calibri.ttf",
            ]
            self.fonts = [f for f in system_fonts if os.path.exists(f)]

        self.aug = get_augmentation_pipeline()

    def generate_image(self):
        w, h = sample_crop_size()
        bg_img = create_background(w, h)
        font_path = random.choice(self.fonts) if self.fonts else None

        # Генерируем оверлей текста (прямой или по дуге)
        text_overlay = render_text_overlay(font_path, w, h)
        text_color = get_contrasting_color(bg_img)

        # Применяем цвет текста к RGBA маске
        colored_text = Image.new("RGBA", text_overlay.size, text_color + (255,))
        colored_text.putalpha(text_overlay.split()[3])

        # Вставляем по центру
        paste_x = (w - text_overlay.width) // 2
        paste_y = (h - text_overlay.height) // 2
        bg_img.paste(colored_text, (paste_x, paste_y), colored_text)

        # Выбор поворота (0 или 180) с вероятностью 50/50 + мелкий джиттер угла
        label_180 = 0
        base_angle = 180.0 if label_180 == 1 else 0.0
        angle = base_angle + random.uniform(-3.0, 3.0)

        # Поворачиваем картинку
        rotated_img = bg_img.rotate(angle, resample=Image.BICUBIC, expand=False, fillcolor=text_color)

        # Аугментации
        img_np = np.array(rotated_img)
        augmented = self.aug(image=img_np)["image"]

        return augmented, label_180, (w, h)


def build_dataset(num_samples=5000, output_dir="synthetic_dataset", fonts_dir=None):
    images_dir = os.path.join(output_dir, "images")
    os.makedirs(images_dir, exist_ok=True)

    generator = SyntheticDatasetGenerator(fonts_dir=fonts_dir)
    records = []

    print(f"Генерация {num_samples} изображений с плотным заполнением текста...")
    for i in range(num_samples):
        img_np, label_180, (w, h) = generator.generate_image()

        image_id = f"synth_{i:06d}.jpg"
        img_path = os.path.join(images_dir, image_id)

        cv2.imwrite(img_path, cv2.cvtColor(img_np, cv2.COLOR_RGB2BGR))

        records.append({
            "image_id": image_id,
            "p_180": label_180,
            "width": w,
            "height": h
        })

        if (i + 1) % 1000 == 0 or (i + 1) == num_samples:
            print(f"Обработано [{i + 1}/{num_samples}]")

    df = pd.DataFrame(records)
    df.to_csv(os.path.join(output_dir, "dataset.csv"), index=False)
    print(f"Готово! Сохранено в '{output_dir}'.")


if __name__ == "__main__":
    build_dataset(num_samples=40000, output_dir="synthetic_dataset", fonts_dir='fonts')
'''
```

```python
'''
import os
from collections import Counter
from concurrent.futures import ThreadPoolExecutor
from pathlib import Path
import numpy as np
from PIL import Image

# 1. Укажите путь к папке с изображениями
IMAGE_DIR = "./synthetic_dataset"
# Допустимые расширения файлов
EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp", ".bmp", ".tiff"}


def get_image_size(file_path):
    """Получает (width, height) изображения без полной загрузки в память."""
    try:
        with Image.open(file_path) as img:
            return img.size  # Возвращает кортеж (ширина, высота)
    except Exception:
        # Игнорируем поврежденные или не являющиеся изображениями файлы
        return None


def main():
    image_folder = Path(IMAGE_DIR)

    # Собираем пути ко всем изображениям
    file_paths = [
        f for f in image_folder.rglob("*") if f.suffix.lower() in EXTENSIONS
    ]

    print(f"Найдено файлов: {len(file_paths)}. Чтение размеров...")

    # Параллельное чтение файлов в многопоточном режиме (ускоряет работу в разы)
    sizes = []
    with ThreadPoolExecutor() as executor:
        results = executor.map(get_image_size, file_paths)
        sizes = [s for s in results if s is not None]

    if not sizes:
        print("Изображения не найдены или не удалось прочитать.")
        return

    # Разделяем ширину и высоту
    widths, heights = zip(*sizes)

    # 2. Средние и медианные значения
    mean_width = np.mean(widths)
    median_width = np.median(widths)

    mean_height = np.mean(heights)
    median_height = np.median(heights)

    # 3. Топ-20 самых популярных размеров
    size_counts = Counter(sizes)
    top_20 = size_counts.most_common(30)

    # Вывод результатов
    print("\n" + "=" * 40)
    print("СТАТИСТИКА РАЗМЕРОВ ИЗОБРАЖЕНИЙ")
    print("=" * 40)
    print(f"Обработано изображений: {len(sizes)}")
    print(
        f"Ширина:   Средняя = {mean_width:.1f}px | Медиана = {median_width:.0f}px"
    )
    print(
        f"Высота:  Средняя = {mean_height:.1f}px | Медиана = {median_height:.0f}px"
    )

    print("\nТоп-30 самых популярных размеров (Ширина x Высота):")
    print("-" * 45)
    for rank, ((w, h), count) in enumerate(top_20, 1):
        percentage = (count / len(sizes)) * 100
        print(f"{rank:2d}. {w}x{h:<12} — {count} шт. ({percentage:.2f}%)")


if __name__ == "__main__":
    main()
'''
```
